```python
#does geography relate to churn
bank_data.groupby('Geography')['Exited'].agg(
    ['count' , 'sum' , 'mean']
).sort_values('mean' , ascending = False)
```

```python
geography_churn = (
    bank_data.groupby('Geography')['Exited']
    .mean()
    .mul(100)
    .sort_values(ascending = False)
)
geography_churn
```

which country has the highest observed customer churn rate

```python
#does age relate to churn?
bank_data.groupby('Age')['Exited'].mean()
```

```python
bins = [0, 25, 35, 45, 55, 65, 100]
labels = [
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    '56-65',
    '66+'
]

bank_data['AgeGroup'] = pd.cut(
    bank_data['Age'],
    bins = bins,
    labels = labels
)
```

```python
age_churn = (bank_data.groupby('AgeGroup', observed = True)['Exited']
             .mean()
             .mul(100)
            ).sort_values(ascending = False)
age_churn
```
