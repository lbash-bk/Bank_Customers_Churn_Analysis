```python

# ==========================================
# BIVARIATE ANALYSIS: CUSTOMER CHURN
# ==========================================

# Create age groups
bins = [0, 25, 35, 45, 55, 65, 100]
labels = ['18-25', '26-35', '36-45', '46-55', '56-65', '66+']

bank_data['AgeGroup'] = pd.cut(
    bank_data['Age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Function for categorical variables
def categorical_churn_analysis(column):
    result = bank_data.groupby(column, observed=False)['Exited'].agg(
        Total_Customers='count',
        Churned_Customers='sum',
        Churn_Rate='mean'
    )
    result['Churn_Rate'] = (result['Churn_Rate'] * 100).round(2)
    print(f"\n{column.upper()} VS CHURN")
    print(result.to_string())


# Function for numerical variables
def numerical_churn_analysis(column):
    result = bank_data.groupby('Exited')[column].agg(
        ['count', 'mean', 'median', 'min', 'max']
    )
    print(f"\n{column.upper()} VS CHURN")
    print(result.round(2).to_string())


print("=" * 60)
print("BIVARIATE ANALYSIS")
print("=" * 60)

# Categorical variables
for column in [
    'Geography',
    'Gender',
    'AgeGroup',
    'HasCrCard',
    'IsActiveMember',
    'NumOfProducts'
]:
    categorical_churn_analysis(column)

# Numerical variables
for column in [
    'Age',
    'CreditScore',
    'Balance',
    'EstimatedSalary',
    'Tenure'
]:
    numerical_churn_analysis(column)

print("\n" + "=" * 60)
print("END OF BIVARIATE ANALYSIS")
print("=" * 60)
```

```python

# ==========================================
# MULTIVARIATE ANALYSIS: CUSTOMER CHURN
# ==========================================

def multivariate_churn_analysis(columns):
    result = bank_data.groupby(
        columns,
        observed=False
    )['Exited'].agg(
        Total_Customers='count',
        Churned_Customers='sum',
        Churn_Rate='mean'
    )

    result['Churn_Rate'] = (
        result['Churn_Rate'] * 100
    ).round(2)

    print(f"\n{' + '.join(columns).upper()} VS CHURN")
    print(result.to_string())


print("=" * 60)
print("MULTIVARIATE ANALYSIS")
print("=" * 60)

combinations = [
    ['Geography', 'Gender'],
    ['Geography', 'IsActiveMember'],
    ['AgeGroup', 'IsActiveMember'],
    ['NumOfProducts', 'IsActiveMember'],
    ['Geography', 'NumOfProducts'],
    ['Gender', 'AgeGroup'],
    ['Geography', 'Gender', 'IsActiveMember'],
    ['AgeGroup', 'Geography', 'IsActiveMember']
]

for combination in combinations:
    multivariate_churn_analysis(combination)

print("\n" + "=" * 60)
print("END OF MULTIVARIATE ANALYSIS")
print("=" * 60)
```

```python

# ==========================================
# CORRELATION ANALYSIS
# ==========================================

numerical_data = bank_data.select_dtypes(include='number')

correlation = numerical_data.corr()

print("CORRELATION WITH CUSTOMER CHURN")
print(
    correlation['Exited']
    .sort_values(ascending=False)
    .round(3)
    .to_string()
)
```
